**Задание № ОДУ-5**

Дана третья краевая задача для ОДУ второго порядка следующего вида:

$$ (k(x)u'(x))' - q(x)u(x) = -f(x), \quad 0 \le x \le 1, $$

$$ k(0)u'(0) = \varkappa_0 u(0) - g_0 $$
$$ -k(1)u'(1) = \varkappa_1 u(1) - g_1 $$

где

| $f(x)$ | $k(x)$ | $q(x)$ | $\varkappa_0$ | $g_0$ | $\varkappa_1$ | $g_1$ |
| :---: | :---: | :---: | :---: | :---: | :---: | :---: |
| $e^x$ | $\sin^2 x + 1$ | $\cos x$ | 1 | 1 | 1 | 1 |

Необходимо:

**1.** Используя разностные операторы вместо дифференциальных, аппроксимировать поставленную задачу разностной схемой 2-го порядка на минимальном шаблоне. Для повышения порядка аппроксимации разностных условий выделить главный член погрешности и заменить его, используя следующий вид исходного дифференциального уравнения: $k(x)u''(x) + k'(x)u'(x) - q(x)u(x) = -f(x)$.

**2.** Интегро-интерполяционным методом построить консервативную разностную схему. Для вычисления коэффициентов этой схемы использовать формулу трапеций.

**3.** Аппроксимировать исходную задачу вариационно-разностным методом. Для вычисления коэффициентов полученной схемы использовать формулу средних прямоугольников.

**4.** Методом разностной прогонки реализовать полученные в п.п. 1-3 разностные схемы при $h = 0.1$. Провести сравнительный анализ сеточных решений при разных шагах.

## Задание 1

$$k(x)u''(x) + k'(x)u'(x) - q(x)u(x) = -f(x),\ \ 0 \le x \le 1$$

$$\overline{w_h} = \{ x_i=ih, i=\overline{0,N}, h=\frac{1}{N} \}$$

$$Ш(x)=\{ x_{i-1}, x_i, x_{i+1} \}$$

$$\begin{cases} k y_{\bar{x} x} + k' y_{\mathring{x}} - qy = -f,\ \ x \in w_h \\
k(0)y_x(0) = \tilde{\varkappa_0} y(0) - \tilde{g_0} \\
-k(1)y_x(1) = \tilde{\varkappa_1} y(1) - \tilde{g_1} \end{cases}$$

Используя невязку разностного уравнения на точном решении в точках $0$ и $1$, зададим значения коэффициентов $\tilde{\varkappa_0}, \tilde{\varkappa_1}, \tilde{g_0}, \tilde{g_1}$ таким образом, чтобы порядок аппроксимации разностной схемы был равен $2$.

$$\begin{gathered} 
\tilde{\varkappa_0} = (1 - \frac{h}{2} \frac{k'(0)}{k(0)}) \varkappa_0 + \frac{h}{2} q(0) \\
\tilde{g_0} = (1- \frac{h}{2} \frac{k'(0)}{k(0)})g_0 + \frac{h}{2} f(0)
\end{gathered}$$

$$\begin{gathered} 
\tilde{\varkappa_1} = (1 + \frac{h}{2} \frac{k'(1)}{k(1)}) \varkappa_1 + \frac{h}{2} q(1) \\
\tilde{g_1} = (1 + \frac{h}{2} \frac{k'(1)}{k(1)})g_1 + \frac{h}{2} f(1)
\end{gathered}$$

$$\begin{cases}
k(x_i) \frac{y(x_{i+1}) - 2y(x_i) + y(x_{i-1})}{h^2} + k'(x_i) \frac{y(x_{i+1}) - y(x_{i-1})}{2h} - q(x_i)y(x_i) = -f(x_i) \\
k(x_0) \frac{y(x_1) - y(x_0)}{h} = \tilde{\varkappa_0} y(x_0) - \tilde{g_0} \\
-k(x_N) \frac{y(x_N) - y(x_{N-1})}{h} = \tilde{\varkappa_1} y(x_N) - \tilde{g_1}
\end{cases}$$

Приведём к трёхдиагональному виду:
$$\begin{cases}
(k_i - \frac{h}{2} k'_i)y_{i-1} - (2k_i +h^2q_i)y_i + (k_i + \frac{h}{2} k'_i)y_{i+1} = -h^2f_i \\
-(k_0 + h \tilde{\varkappa_0})y_0 + k_0 y_1 = -h \tilde{g_0} \\
k_N y_{N-1} - (k_N + h \tilde{\varkappa_1}) y_N = -h \tilde{g_1}
\end{cases}$$

где $k_i = k(x_i), k'_i = k'(x_i), q_i = q(x_i), f_i = f(x_i)$

## Задание 2

$$\begin{cases}
(ay_{\bar{x}})_x -dy = -\varphi \\
a_1 y_{x,0} = (\varkappa_0 + \frac{h}{2} d_0)y_0 - (g_0 + \frac{h}{2} \varphi_0) \\
-a_N y_{\bar{x},N} = (\varkappa_1 + \frac{h}{2} d_N) y_N - (g_1 + \frac{h}{2} \varphi_N)
\end{cases}$$
где: $$a_i = (\frac{1}{h} \int_{x_{i-1}}^{x_i} \frac{1}{k(x)} dx)^{-1},\ d_i=\frac{1}{h} \int_{x_{i-\frac{1}{2}}}^{x_{i+\frac{1}{2}}} q(x)dx,\ \varphi_i =\frac{1}{h}  \int_{x_{i-\frac{1}{2}}}^{x_{i+\frac{1}{2}}} f(x)dx$$
$$d_0 = \frac{2}{h} \int_0^{\frac{h}{2}} q(x)dx,\ \varphi_0 = \frac{2}{h} \int_0^{\frac{h}{2}} f(x)dx,\ d_N = \frac{2}{h} \int^1_{1 - \frac{h}{2}} q(x)dx,\ \varphi_N = \frac{2}{h} \int^1_{1 - \frac{h}{2}} f(x)dx$$

Для приближённого вычисления интегралов воспользуемся формулой трапеций: $$\int_a^b f(x) dx \approx \frac{f(a) + f(b)}{2} \cdot (b-a)$$

Таким образом, $$a_i = \frac{2 k(x_i) k(x_{i-1})}{k(x_{i-1}) + k(x_i)},\ d_i = \frac{q(x_{i+\frac{1}{2}}) + q(x_{i-\frac{1}{2}})}{2},\ \varphi_i = \frac{f(x_{i+\frac{1}{2}}) + f(x_{i-\frac{1}{2}})}{2}$$
$$d_0 = \frac{q(\frac{h}{2})+q(0)}{2},\ \varphi_0 = \frac{f(\frac{h}{2})+f(0)}{2},\ d_N = \frac{q(1)+q(1-\frac{h}{2})}{2},\ \varphi_N = \frac{f(1)+f(1-\frac{h}{2})}{2}$$

$$\begin{cases}
\frac{1}{h} (a_{i+1} \frac{y_{i+1} - y_i}{h} - a_i \frac{y_i-y_{i-1}}{h}) - d_i y_i = - \varphi_i \\
a_1 \frac{y_1-y_0}{h} = (\varkappa_0 + \frac{h}{2} d_0)y_0 - (g_0 + \frac{h}{2} \varphi_0) \\
-a_N \frac{y_N-y_{N-1}}{h} = (\varkappa_1 + \frac{h}{2} d_N) y_N - (g_1 + \frac{h}{2} \varphi_N)
\end{cases}$$

Приведём к трёхдиагональному виду:

$$\begin{cases}
a_i y_{i-1} - (a_i + a_{i+1} + h^2d_i)y_i + a_{i+1}y_{i+1} = -h^2 \varphi_i \\
-(a_1 + h \varkappa_0 + \frac{h^2}{2} d_0) y_0 + a_1 y_1 = -h(g_0 + \frac{h}{2} \varphi_0) \\
a_N y_{N-1} - (a_N + h \varkappa_1 + \frac{h^2}{2} d_N)y_N = -h(g_1 + \frac{h}{2} \varphi_N)
\end{cases}$$

$$\begin{cases}

$$
