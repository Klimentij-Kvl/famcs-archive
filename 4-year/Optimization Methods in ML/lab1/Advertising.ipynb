```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
df = pd.read_csv("Advertising.csv")
df.head()
```

```python
df = df.drop("Unnamed: 0", axis=1)
df.head()
```

```python
X = df.drop("sales", axis=1).to_numpy()
X[:5]
```

```python
y = df["sales"].to_numpy()
y[:5]
```

```python
w_true = np.linalg.inv(X.T @ X) @ X.T @ y
w_true
```

```python
model = LinearRegression().fit(X, y)
w_est = model.coef_
w_est
```
